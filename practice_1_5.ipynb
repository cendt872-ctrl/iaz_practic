```python
n = 5  # <-- ЗАМІНІТЬ 5 на ваш номер у списку журналу!
limit = n + 10

print(f"--- Цикл for (від 1 до {limit}) ---")
for i in range(1, limit + 1):
    print(i)

print(f"\n--- Цикл while (від {limit} до 1) ---")
j = limit
while j >= 1:
    print(j)
    j -= 1
```
